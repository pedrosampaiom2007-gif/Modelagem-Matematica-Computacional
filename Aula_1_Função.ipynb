###3) Pra f(x) = 3x - 1, gere 100 pontos em [−2, 2], plote e marque o zero.


```python
#Pra f(x) = 3x - 1, gere 100 pontos em [−2, 2], plote e marque o zero.
x = np.linspace(-2, 2, 100)
y = 3 * x - 1

# Calculate the x-intercept (where y=0)
x_zero = 1 / 3

plt.figure(figsize=(10,6)) # Correção: usar figsize=(largura, altura)
plt.plot(x,y,color="red", label='f(x) = 3x - 1') # Adicionado label para a legenda
plt.title("Gráfico da função f(x) = 3x - 1 ")

# Correção: Adicionado o parâmetro 'y' (0) e formatado o label com f-string
plt.scatter(x_zero, 0, color='blue', label=f'Zero={x_zero:.2f}', s=100, edgecolors='black')

# Adicionando de volta as linhas de eixo e grade para clareza
plt.axhline(0, color='green', linestyle='--', linewidth=0.8)
plt.axvline(0, color='yellow', linestyle='--', linewidth=0.8)

plt.xlabel('x') # Adicionado rótulo para o eixo x
plt.ylabel('f(x)') # Adicionado rótulo para o eixo y
plt.grid(True) # Adicionado grade
plt.legend() # Adicionado legenda
plt.show() # Para exibir o gráfico
```

###4) Escreva my_abs com np.where e teste com -6.54 e com np.array([-3, 0, 2]). Depois tente com if e veja o erro.


```python
def my_abs(x):
    return np.where(x >= 0, x, -x)
```

```python
my_abs(-6.54)
```

```python
my_abs(np.array([-3, 0, 2]))
```

```python
def my_abs_if(x):
    if x >= 0:
        return x
    return -x

my_abs_if(np.array([-3, 0, 2]))
```

##5) Faça f(x) = x² para x < 0 e x + 1 para x ≥ 0, com np.where, e plote.


```python
def ex(x):
    return np.where(x < 0, x**2, x + 1)

y5 = ex(x)  # x já existe, do exercício 3

plt.plot(x, y5, color="blue")
plt.axvline(0, color="gray", linestyle="--", linewidth=0.8)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("f(x) = x² se x<0, x+1 se x≥0")
plt.show()
```

## 6) Função do discriminante Δ = b² − 4ac que aceita arrays. Teste com a=1, b=[−4, 2, 0], c=4 e diga qual raiz cada caso tem.


```python
def delta(a,b,c):
  # Converte b e c para arrays numpy para permitir operações elemento a elemento
  b_array = np.asarray(b)
  c_array = np.asarray(c)

  conta = b_array**2 - 4*a*c_array

  return conta
```

```python
delta(1,[-4, 2, 0],4)
#Δ=0 (b=−4): uma raiz real (raiz dupla)
#Δ=−12 (b=2): nenhuma raiz real (duas raízes complexas)
#Δ=−16 (b=0): nenhuma raiz real (duas raízes complexas)
```
